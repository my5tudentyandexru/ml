## 4. Jupyter: как устроена тетрадь

Тетрадь — это список ячеек двух типов:
- **Code** — код, выполняется ядром (kernel); результат последнего выражения выводится под ячейкой;
- **Markdown** — текст с разметкой: заголовки, списки, таблицы, формулы, картинки.

### Горячие клавиши (режим команд — `Esc`, режим редактирования — `Enter`)
| Клавиши | Действие | Colab |
|---|---|---|
| `Shift+Enter` | выполнить ячейку и перейти к следующей | так же |
| `Ctrl+Enter` | выполнить и остаться | так же |
| `A` / `B` | вставить ячейку выше / ниже | `Ctrl+M A` / `Ctrl+M B` |
| `D D` | удалить ячейку | `Ctrl+M D` |
| `M` / `Y` | сделать ячейку Markdown / Code | `Ctrl+M M` / `Ctrl+M Y` |
| `Shift+Tab` | подсказка по функции под курсором | наведение мыши |
| `Tab` | автодополнение | так же |

### Главная ловушка Jupyter: скрытое состояние

Ячейки можно запускать в **любом** порядке и **сколько угодно** раз. Переменные живут в памяти ядра,
пока его не перезапустили. Поэтому тетрадь может «работать» у вас и падать у всех остальных:
вы когда-то выполнили ячейку, которую потом удалили, а переменная осталась в памяти.

Запустите следующую ячейку **три раза подряд** и посмотрите на результат.

````python
balance = 1000
````

````python
balance = balance * 1.1       # начисляем 10%: запустите эту ячейку несколько раз
round(balance, 2)
````

Номер в квадратных скобках слева от ячейки (`[5]`) — порядок выполнения. Если номера идут не подряд,
тетрадь выполнялась не сверху вниз.

**Правило:** перед тем как отправить тетрадь, выполните **Kernel → Restart & Run All**
(в Colab: **Среда выполнения → Перезапустить сеанс и выполнить все**).
Если тетрадь дошла до конца без ошибок, её сможет выполнить и другой человек.

## 5. Magic-команды и справка

Magic-команды — возможности Jupyter (точнее, IPython), а не языка Python. В обычном `.py`-файле они не работают.

| Команда | Что делает |
|---|---|
| `%timeit выражение` | многократно замеряет время выполнения строки |
| `%%time` | один раз замеряет время всей ячейки (пишется первой строкой) |
| `%who`, `%whos` | какие переменные сейчас в памяти ядра |
| `%%writefile имя.py` | записывает содержимое ячейки в файл |
| `%run имя.py` | запускает скрипт |
| `%pip install пакет` | ставит пакет в окружение тетради |
| `!команда` | выполняет команду терминала (`!ls`, `!nvidia-smi`) |
| `функция?` | справка по объекту; `??` — ещё и исходный код |

````python
numbers = list(range(1_000_000))
%timeit sum(numbers)
````

````python
%%time
squares = [x ** 2 for x in numbers]
len(squares)
````
