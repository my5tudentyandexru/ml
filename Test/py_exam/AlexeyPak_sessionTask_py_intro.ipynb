Задание 2 (6 баллов)
Сжатие строки

Напишите функцию, которая принимает строку и сжимает её определённым образом. Строки содержат произвольные символы, включая пробелы и спецсимволы, и требуют точного учета длины при кодировании. Сжатие строки происходит сериями одинаковых символов в формате символколичество, но только если длина сжатой строки не превышает исходную.

Примеры:
вход: aaabbc → вывод: a3b2c,
вход: abcd → вывод: abcd (так как сжатая равна).

Критерии:
Правильная реализация логики сжатия — 3 балла
Учет условия длины — 2 балла
Обработка краевых случаев (пустые строки, одиночные символы) — 1 балл

```python
# Различные значения для проверки
# string = " 8312ha21jdw  "
string = "  aaaddd   66b  *32d  "

def transform(string: str) -> str:
    # Очистка строки (Строки содержат произвольные символы, включая пробелы и спецсимволы)
    clean_text = "".join(ch for ch in string if ch.isalnum()) # join - сшиваем (isalnum - только буквенно‑цифровые символы)
    # защита от пустой строки
    if not clean_text:  
        return ""
    result = []
    count = 1
    for i in range(1, len(clean_text)):                  # повторять в цикле с каждым элементом индекс для i (от 1 до количества элементов чистой строки)
        if clean_text[i] == clean_text[i - 1]:           # если элемент чистой строки равен предыдущему элементу чистой строки
            count += 1                                   # то кол-во увеличивается на 1
        else:                                            # в оставшихся случаях 
            # добавить в список (символ и количество, если оно больше 1)
            result.append(f"{clean_text[i - 1]}{count if count > 1 else ''}")
            count = 1                                    # обновить количество элементов
    # добавить в список последний символ и его количество (если оно больше 1)
    result.append(f"{clean_text[-1]}{count if count > 1 else ''}")
    encoded = "".join(result)                            # сшить элементы списка в одну строку
    # если длина сжатой строки не превышает длину очищенной — возвращаем сжатую, иначе очищенную
    return encoded if len(encoded) <= len(clean_text) else clean_text


# Проверка
# вход: aaabbc → вывод: a3b2c,
# вход: abcd → вывод: abcd (так как сжатая равна).
print(transform(string))
```

Задание 3 (10 баллов)
Проверка скобок

Напишите функцию, которая проверяет корректность расстановки круглых, квадратных и фигурных скобок.

Критерии:
Распознавание всех трёх типов скобок — 3 балла
Учёт вложенности — 4 балла
Чистота кода и читаемость — 2 балла
Возврат булевого результата — 1 балл

```python
# Комментарий: данное задание было в качестве домашнего задания и было сделано с преподавателем при разборе домашнего задания

# Различные значения для проверки
# text = "([({  })()])" # True
text = "([({ ( })()])"  # False

PAIRS = {")": "(", "]": "[", "}": "{"}
OPEN = set(PAIRS.values())
CLOSE = set(PAIRS.keys())


def find_unbalanced(text: str) -> int:
    stack = []
    for i, ch in enumerate(text):
      if ch in OPEN:
        stack.append((i, ch))
      elif ch in CLOSE:
        if not stack or stack[-1][1] != PAIRS[ch]:
          return i
        stack.pop()
    if stack:
      return stack[0][0]
    return -1

def check_brackets(text: str) -> bool:
    return find_unbalanced(text) == -1

# Проверка
print(check_brackets(text))
```

Задание 4 (10 баллов)
Генератор паролей
Напишите функцию, которая генерирует пароль заданной длины. В реализации надо учитывать, что:

Пароль должен содержать буквы, цифры и специальные символы.
Длина пароля задается пользователем.

Критерии:
Корректная генерация длины — 3 балла
Наличие всех типов символов — 4 балла
Уникальность и случайность паролей — 2 балла
Удобство использования функции — 1 балл

```python
import random

# Напишите функцию, которая генерирует пароль заданной длины.
def gen_pass():
    pwd_len = None # длинна пароля
    count = 5      # количество попыток ввода
    while count > 0:
        user_input = input("Введите длину пароля (целое число ≥ 3): ") # Длина пароля задается пользователем. >= 3, т.к. должен содержать хотя бы одну букву, цифру и специальный символ
        # Проверка на корректность ввода
        if user_input.isdigit() and int(user_input) >= 3:              
            pwd_len = int(user_input)
            # сшиваем (рендом по 1 символу из каждого набора + дополняем рендом с разрешенными повторами из chars в количестве user_input - 3)
            pwd = "".join(random.choice(letters))\
                + "".join(random.choice(digits))\
                + "".join(random.choice(specials))\
                + "".join(random.choices(chars, k=pwd_len - 3))
            # еще один рендом, т.к. выше сшивали по порядку
            pwd = "".join(random.sample(pwd, k=len(pwd)))              # sample - случайная перестановка
            break                                                      # длинна введена корректно, пароль сформирован, завершаем цикл.
        else:                                                          # иначе
            count -= 1                                                 # сократить количество попыток ввода длины пароля на 1 
            print(f"Введите длину пароля (целое число ≥ 3). Осталось попыток ввода: {count}")

    if pwd_len is not None:
        print("Сгенерированный пароль:", pwd, sep="\n")
    else:
        print("Длинна пароля введена некорректно")

# Пароль должен содержать буквы, цифры и специальные символы.
letters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ"
digits = "0123456789"
specials = "!@#$%&-_+="
chars = letters + digits + specials

# Проверка
gen_pass()
```

Задание 5 (6 баллов)
Почти палиндром

Напишите программу, которая проверяет, является ли строка палиндромом или «почти палиндромом». «Почти палиндром» означает, что можно удалить одну букву, чтобы строка стала палиндромом.

Критерии:
Проверка палиндромов — 3 балла
Обработка случая «почти палиндром» — 2 балла
Учет регистра и пробелов — 1 балл

```python
# Комментарий: данное задание было в качестве домашнего задания и было сделано с преподавателем при разборе домашнего задания

# Различные значения для проверки
# s ="А роза упала на лапу Азора" # True
# s ="120500404005021" # True
s ="12050440404005021" # False

def normalize(s):
    """Оставляет только буквы и цифры, приводит к нижнему регистру."""
    return "".join(ch.lower() for ch in s if ch.isalnum())


def is_palindrome(s):
    """True, если нормализованная строка читается одинаково в обе стороны."""
    clear = normalize(s)
    return clear == clear[::-1]


def is_almost_palindrome(s):
    """True, если строка станет палиндромом после удаления не более одного символа."""
    clean = normalize(s)
    left_idx, right_idx = 0, len(clean) - 1
    while left_idx < right_idx:
      if clean[left_idx] != clean[right_idx]:
        without_left = clean[:left_idx] + clean[left_idx+1:]
        without_right = clean[:right_idx] + clean[right_idx+1:]
        return is_palindrome(without_left) or is_palindrome(without_right)

      left_idx += 1
      right_idx -= 1
    return True

# Проверка
print(is_almost_palindrome(s))
```

Задание 6 (8 баллов)
Задача Шредингера

Разработайте программу, которая «стирает» фрагменты текста в файле.

Например:
Пользователь указывает файл и процент текста, который нужно удалить (например, 30%).
Программа случайно выбирает слова или части абзацев и заменяет их на пробел или ..., сохраняя общий объем документа.
Вход:
«Сегодня солнечный день, и я собираюсь гулять в парке с моими друзьями.»
Вывод (удалено ~30%):
«Сегодня ... день, и я собираюсь гулять ... моими друзьями.»


Критерии:
Чтение и анализ файла — 2 балла
Алгоритм случайного удаления — 4 балла
Запись результата — 1 балл
Корректный ввод параметров пользователем — 1 балл

```python
# Для читаемости создаем 5 функций
# 1) Ввод процента и проверка на корректность ввода
# 2) Ввод адреса файла, его открытие или использование по умолчанию
# 3) Замена текста
# 4) Запись результата в файл text_replaced.txt
# 5) Агрегирующая функция

import random
import re #  для разбивки строки по шаблону (регулярному выражению)

# 1) Ввод процента и проверка
def get_percent():
    count = 3 # количество попыток ввода, если не число, в задании не указано int или float = используем float
    while count > 0:
        input_percents = input("Введите целое число от 0 до 100: ")
        # Проверка на корректность ввода
        try:
            value = float(input_percents)
            if 0 <= value <= 100:
                return value
            else:
                count -= 1
        except ValueError:
            count -= 1
        if count > 0:
            print(f"Введите число от 0 до 100. Осталось попыток ввода: {count}")
    # Если цикл завершился при превышении попыток, input_percents может быть некорректным, то возвращаем значение только если оно валидно, иначе None.
    return None

# 2) Ввод адреса файла, его открытие или использование по умолчанию
def read_file_text(file_path="text.txt"):
    # путь к файлу задаётся file_path (по умолчанию "text.txt")
    file_path = input("Введите путь до файла (Enter — text.txt): ").strip()
    try:
        with open(file_path, mode='r', encoding='utf-8') as f:
            text = f.read()
    except FileNotFoundError:
        print(f"Файл '{file_path}' не найден, используем путь по умолчанию: text.txt")
        file_path = "text.txt"  # если файла нет, используется путь по умолчанию
        with open(file_path, mode='r', encoding='utf-8') as f:
            text = f.read()
    return text

# 3) Замена текста
def replace_text(text, input_percents, replacement="..."):
    # Используем регулярное выражение, чтобы разделить текст на слова и знаки препинания и пробелы
    list_elems = re.split(r"(\w+)", text) #  разбивает строку по шаблону (регулярному выражению).
    # Находим индексы всех элементов, которые являются словами
    elem_idx = [i for i, elem in enumerate(list_elems) if elem.isalnum()]
    if not elem_idx:
        return text
    # Вычисляем количество слов для удаления
    elems_remove_count = int(len(elem_idx) * (input_percents / 100))
    # Случайно выбираем индексы слов, которые будут заменены
    idx_replace = random.sample(elem_idx, elems_remove_count)
    # Заменяем выбранные слова на replacement="..."
    for idx in idx_replace:
        list_elems[idx] = replacement
    # Сшиваем текст в единую строку
    return "".join(list_elems)

# 4) Запись результата в файл text_replaced.txt
def save_processed_text(text, input_percents):
    processed_text = replace_text(text, input_percents)
    with open("text_replaced.txt", "w", encoding="utf-8") as f:
        f.write(processed_text)
    return processed_text

# 5) Агрегирующая функция, связывает все функции
def main():
    # 1) Получаем процент от пользователя
    input_percents = get_percent()
    if input_percents is None:
        print("Превышено количество попыток.")
        return
    # 2) Читаем текст из файла
    try:
        text = read_file_text("text.txt")
    except FileNotFoundError:
        print("Файл text.txt не найден. Положите его рядом со скриптом.")
        return
    # 3) Заменяем слова сохраняем результат
    processed_text = save_processed_text(text, input_percents)
    print(processed_text)                           # для проверки вывод результата, записываемого в файл
    print(f"{input_percents}% слов, результат сохранен в text_replaced.txt")

main()
```
